## Scenario & Dataset

Imagine we’re studying how students’ **exam scores** depend on:

- `StudyHours`: hours of study per week  
- `SleepHours`: average hours of sleep per night  
- `PrevGPA`: previous semester GPA  

We’ll use a small synthetic dataset and walk through z-scores, simple linear regression,
multiple linear regression, and MSE.


```python

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

data = pd.DataFrame({
    "StudyHours": [5, 10, 3, 8, 15, 12, 4, 9, 7, 11],
    "SleepHours": [6.0, 7.5, 5.5, 7.0, 8.0, 6.5, 5.0, 7.0, 6.0, 7.5],
    "PrevGPA":    [2.5, 3.0, 2.2, 2.8, 3.5, 3.2, 2.4, 3.0, 2.7, 3.3],
    "ExamScore":  [60, 75, 55, 70, 90, 82, 58, 78, 68, 80]
})

data
```


## Part 1 – Z-Scores

The **z-score** of a value \(x\) is defined as:

$$z = \frac{x - \mu}{\sigma}$$

where mu is the mean and sigma is the standard deviation.

**Q1:**  
Compute the z-scores for `StudyHours` to see how far each student is from the average. Add a new column named `StudyHours_z` to `data` containing those z-scores.


```python
mu = data["StudyHours"].mean()
sigma = data["StudyHours"].std()

data["StudyHours_Z"] = (data["StudyHours"] - mu) / sigma

display(data)
```


### Interpreting Z-Scores

**Q2:**  
- If a student has `StudyHours_z = 1.5`, what does that mean in words?  
- Are there any students more than 2 standard deviations above or below the mean?


```python

q2_answer = """
A StudyHours_Z of 1.5 means the student studies 1.5 standard deviations above the average,
so they study much more than the typical student in the dataset.

No, there are no students more than 2 standard deviations above or below the mean.
"""
print(q2_answer)
```


## Part 2 – Simple Linear Regression (ExamScore ~ StudyHours)

We start with a simple model that uses only `StudyHours` to predict `ExamScore`:

$$
\text{ExamScore} = \beta_0 + \beta_1 \cdot \text{StudyHours}
$$

**Q1:**  

Implement a simple linear regression, and print parameters.

```python
X = data[["StudyHours"]]
y = data["ExamScore"]

model_simple = LinearRegression()
model_simple.fit(X, y)

intercept_simple = model_simple.intercept_
coef_simple = model_simple.coef_[0]

print("Intercept (beta_0):", intercept_simple)
print("Coefficient for StudyHours (beta_1):", coef_simple)
```


### Interpreting the Slope

**Q2:**  
If beta_1 is, for example, around 2.5–3.0, what does that mean in the context of this problem?


```python

q2_answer = f"""
A beta_1 value around 2.5–3.0 means that for each additional hour a student studies, their exam score is predicted to increase by about 2.5 to 3 points on average.
"""
print(q2_answer)
```


## Part 3 – Predictions and MSE (Simple Model)

We now use the fitted model to make predictions and compute the **Mean Squared Error (MSE)**:

$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^n \left(y_i - \hat{y}_i\right)^2
$$

Lower MSE indicates that the model's predictions are, on average, closer to the true values.

1. Add a new column named `ExamScore_pred_single` to `data`. This new column should contain those predicted exam scores.
2. compute MSE using sklearn
3. Manual MSE for the first 3 data points without using sklearn.


```python
# 1
data["ExamScore_pred_simple"] = model_simple.predict(X)

# 2
mse_simple = mean_squared_error(data["ExamScore"], data["ExamScore_pred_simple"])

display(data)
print("MSE (simple model):", mse_simple)

# 3
y_true_3 = data["ExamScore"].iloc[:3]
y_pred_3 = data["ExamScore_pred_simple"].iloc[:3]

mse_3_manual = ((y_true_3 - y_pred_3)**2).mean()

print("Manual MSE for first 3 points:", mse_3_manual)
```


## Part 4 – Multiple Linear Regression

Now we use **all three features**:

- `StudyHours`
- `SleepHours`
- `PrevGPA`

Our model becomes:

$$
\text{ExamScore}
= \beta_0 
+ \beta_1 \cdot \text{StudyHours}
+ \beta_2 \cdot \text{SleepHours}
+ \beta_3 \cdot \text{PrevGPA}
$$

1. Fit this multiple linear regression model and print all coefficients.
2. Predict exam scores by using this multiple linear regression model, and add your results as a new column named `ExamScore_pred_multi` to `data`.
3. Compute MSE for the multiple linear regression model. Show results.


```python
# 1
X_multi = data[["StudyHours", "SleepHours", "PrevGPA"]]
y = data["ExamScore"]

model_multi = LinearRegression()
model_multi.fit(X_multi, y)

intercept_multi = model_multi.intercept_
coefs_multi = model_multi.coef_

print("Intercept (beta_0):", intercept_multi)
print("Coefficients (beta_1, beta_2, beta_3):", coefs_multi)

# 2
data["ExamScore_pred_multi"] = model_multi.predict(X_multi)

# 3
mse_multi = mean_squared_error(data["ExamScore"], data["ExamScore_pred_multi"])

display(data)
print("MSE (multiple model):", mse_multi)
```


### Comparing Model Performance

**Q4:**  
Compare the MSE of the simple model and the multiple model. Which one performs better (has lower MSE)? Why might that be the case?


```python

q4_answer = f"""
The multiple linear regression model has the lower MSE, so it performs better than the simple model.
That makes sense because it uses more information (like StudyHours, SleepHours, and PrevGPA), 
so it can explain more of hte variation in exam scores instead of considering only StudyHours.
"""
print(q4_answer)
```


## Part 5 – Z-Scores of Residuals

Residuals are defined as:

$$
r_i = y_i - \hat{y}_i
$$

1. Compute **z-scores of residuals** to detect large errors. 
2. Are there any residual z-scores with absolute value > 2? What might that indicate?


```python
data["residual_multi"] = data["ExamScore"] - data["ExamScore_pred_multi"]

# 1
mu_res = data["residual_multi"].mean()
sigma_res = data["residual_multi"].std()

data["residual_z"] = (data["residual_multi"] - mu_res) / sigma_res

display(data)

# 2.
large_residuals = data[ data["residual_z"].abs() > 2 ]
print("Residuals with |z| > 2:")
print(large_residuals)
```

```python
q5_answer = f"""
There is one residual z-score with absolute value greater than 2.
This indicates that the model made an unusually large prediction error for that student.
A residual z-score > 2 means the model struggled to predict that exam score accurately and the point may be an outlier.
"""
print(q5_answer)
```
